**Exercise:** A data logger set up to record weather conditions records its readings as text.

The air pressure is `"101325"` (in Pa) and the air temperature is `"15.0"` (in °C).

1. Store both readings as strings, and use `type()` to confirm their type.
2. Convert the readings to floats, and convert the temperature to Kelvin using $T = T_C + 273.15$.
3. Compute the air density using the ideal gas law, $\rho = p/(RT)$, with $R = 287.05$ J/(kg K).
4. Print the density to 3 decimal places using an f-string.
5. Explain why `int("15.0")` raises an error, whereas `float("15.0")` does not.

````python
pressure_str = "101325"
temperature_str = "15.0"
print(type(pressure_str), type(temperature_str))

pressure = float(pressure_str)
temperature = float(temperature_str) + 273.15

gas_constant = 287.05
density = pressure / (gas_constant * temperature)

print(f"Air density: {density:.3f} kg/m^3")
````

`int()` can only convert a string that contains an integer. The decimal point in `"15.0"` tells Python that the value is not an integer, and causes a `ValueError`, whereas `float()` does not.

### Exercise 4
`Mathematical functions`, `Precision`, `f-strings`

A force of $F = 850$ N acts at an angle of $\theta = 35°$ to the horizontal.

1. Compute its components, $F_x = F\cos\theta$ and $F_y = F\sin\theta$, and print them to 2 decimal places.
2. Recover the magnitude from the components using $\texttt{mag} = \sqrt{F_x^2 + F_y^2}$, and print it.
3. Use `magnitude == force` to compare the values and print the result.
3. Use `isclose` from the `math` library to check that this magnitude matches $F$.
4. Explain why `isclose` has to be used here, .

*Hint:* The trigonometric functions in `math` expect radians. Use `mt.radians()` to convert from degrees.

### Solution

````python
force = 850
angle = mt.radians(35)

force_x = force * mt.cos(angle)
force_y = force * mt.sin(angle)
print(f"Fx = {force_x:.2f} N")
print(f"Fy = {force_y:.2f} N")

magnitude = mt.sqrt(force_x**2 + force_y**2)
print(f"Magnitude: {magnitude}")
print(f"Matches F? (exact) {magnitude == force}")
print(f"Matches F? (floating-point) {mt.isclose(magnitude, force)}")
````
