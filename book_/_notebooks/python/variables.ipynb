## What are Variables?

***Variables*** are named containers in which data values are stored. You are already aware of the existence of variables, and use them constantly. For example, observe this algebraic equation:

$$
z = x + y
$$

In this case, $x$, $y$ and $z$ are variables: once the values of $x$ and $y$ are known, the value of $z$ is computed from the above expression. The distinction made in programming is that a variable can hold any ***type*** of data: a whole number, a decimal number, a complex number, a piece of text, a list of measurements, a matrix, and so on.

When you **assign a value to a variable**, Python stores that value in memory and attaches the variable's name to it. Using the name later in the program retrieves the stored value.

Variables in Python are created once a value is assigned to them: <br>

**variable  = value**

So, the assignment is done by $=$. Use `print()` whenever you want to display the value of a variable.

### Example:

```python
x = 2  # assign an integer to x
y = 4  # assign an integer to y
a = "Python is easy!"  # assign a string to a

print(x + y)
print(a)
```

## Naming convention

When writing code, it is recommended that you not only use comments (in Python these start with `#`), but that your code itself is self-descriptive and easy to read. One way to achieve this is to use variable names which describe the purpose of a variable without being too long. Some naming rules are *required* by Python (breaking them causes an error), while others are *recommendations* taken from the [PEP 8 Python style guide](https://peps.python.org/pep-0008/):

1. A variable name may **only contain letters, numbers and the underscore** (`_`) symbol (required by Python).

```python
example_variable_1 = 5
print(example_variable_1)
```

The following name is not allowed, because it contains `!` (this cell will raise a syntax error):

```python
example_variable_1! = 5
print(example_variable_1!)
```

2. A variable name **must start with a letter or an underscore, never a number** (required by Python).

```python
_example_variable_2 = 6
print(_example_variable_2)
```

The following cell will raise a syntax error:

```python
1_example_variable_3 = 7
print(1_example_variable_3)
```

3. Variable names are **case-sensitive**: `manchester` and `Manchester` are two different names (a feature of Python).

```python
manchester = "University of Manchester"   
print(manchester)
```

```python
print(Manchester)                         # you print an undefined variable -> error
```

4. A variable name **cannot be one of Python's reserved keywords**, such as `and`, `if`, `for`, `class`, etc. (required by Python).

```python
and = 9
print(and)
```

You can print the full list of reserved keywords:

```python
import keyword
print(keyword.kwlist)
```

A related trap: Python *does* allow you to use the names of built-in functions, such as `print`, `sum`, `max`, `list` or `type`, as variable names. Doing so hides the original function, which causes confusing errors later on. **Avoid these names**.

```python
sum = 10                  # Allowed, but this hides Python's built-in sum() function
print(sum([1, 2, 3]))     # TypeError: 'int' object is not callable
```

```python
del sum                   # Delete our variable, so that the built-in sum() works again
print(sum([1, 2, 3]))
```

5. The variable name is recommended to be **descriptive**, yet still reasonably short (recommended).

```python
e = 2.71828                     # Too short: what does e mean?
euler = 2.71828                 # Good
the_euler_constant = 2.71828    # Too long
```

6. Use one naming style consistently (recommended). There are several common styles: `snake_case`, `camelCase`, `PascalCase` and `UPPER_CASE`. PEP 8 recommends **`snake_case`** for variable names in Python. By convention, `PascalCase` is reserved for class names and `UPPER_CASE` for constants, so avoid using them for ordinary variables.

```python
man_uni = "University of Manchester"   # snake_case: recommended for variables in Python
manUni = "University of Manchester"    # camelCase: common in other languages (e.g. MATLAB, Java)
ManUni = "University of Manchester"    # PascalCase: used for class names, not variables
SPEED_OF_SOUND = 343                   # UPPER_CASE: used for constants
```

## Creating variables

When we write `pi = 3.14159`, we create (or *declare*) the variable `pi`. Python has no separate declaration step: a variable comes into existence the first time a value is assigned to it. You write the variable name, an equals sign, and then the value.

Python also lets you assign the same value to several variables at once, or assign different values to several variables in a single line. If you need a variable but do not have a value for it yet, you can assign it the special value `None`, which means "no value". Note that you cannot use a variable before it has been assigned: this raises a `NameError`, as you saw with `Manchester` above.

#### Example:

```python
colour = "blue"
print(colour)
```

```python
colour = shirt = "blue"         # Same value for both variables
print(colour, shirt)
```

```python
colour, shirt = "blue", "polo"  # Different values in a single line
print(colour, shirt)
```

```python
empty_variable = None
print(empty_variable)
```

## Variable assignment & manipulation

Technically, a variable is simply a name that refers to an object stored in memory, like a label that points to where the actual data is stored. A variable can be reassigned to a different value at any time. The objects that variables refer to may be immutable (like numbers and strings) or mutable (like lists), which we will come back to later, but the variables themselves can always be reassigned.

```python
colour = "blue"
print(colour)

colour = "red"
print(colour)
```

### `=` means assignment, not equality

In maths, $x = x + 1$ has no solution. In Python, `=` is the **assignment** operator and means: *work out the value on the right-hand side, then store the result in the name on the left*. So `x = x + 1` simply increases `x` by one. To test whether two values are equal, Python uses `==` instead.

```python
x = 10   # Declaring x by assigning a value to it
print("x = ", x)

x = x + 1   # Take the current value (10), add 1 to it, store the result back in x
print("x + 1 = ", x)

x += 1            # Shorthand for x = x + 1
print("x + 1 = ", x)
```

**Note:** 
* $=$ is used for assigning a value to a variable. 
* $==$ check if two values are equal; it returns `True` or `False`

```python
print(x == 12)    # == checks if x is equal to 12
```

Variables can be used in operations such as mathematical calculations or string manipulation. Variables can also be deleted using the `del` statement (as we did with `sum` above). However, this is rarely necessary in Python, as it has a robust "garbage collector" that automatically frees memory used by objects that are no longer referred to by any variable.

## Data (variable) types

In programming, data types are a very important concept. A data type specifies what kind of value something is, and therefore which operations make sense on it. For example, you can divide two numbers, but not two pieces of text.

Python is **dynamically typed**: the type belongs to the **value**, not to the variable. You never state a type when creating a variable, and the same variable can hold an integer on one line and a string on the next. This makes code quick to write, but it also lets some mistakes go unnoticed until the program runs (see [The perils of dynamic typing](#The-perils-of-dynamic-typing) below).


To get the type of a variable/value, use the `type()` function. 

```python
x = 42
print(type(x))       # type() tells us the type of a value (more on this below)
```

```python
x = "forty-two"      # The same variable now holds a string
print(type(x))
```

Here are the fundamental Python data types you'll use most often:

1. **Numeric types:**
    - `int` (integer): whole numbers like `5`, `-17`, `1000`
    - `float` (floating-point): decimal numbers like `3.14`, `-0.001`, `2.0`, or `2.1e5` (scientific notation for $2.1 \times 10^5$)
    - `complex`: complex numbers like `3+4j` (Python uses `j` for the imaginary unit, as is common in engineering)

2. **Text type:**
    - `str` (string): text such as `"Hello"` or `'Python'` (you can use either double or single quotes)

3. **Boolean type:**
    - `bool`: the logical values `True` and `False`. In Python, `bool` is a special kind of `int`: `True` behaves like `1` and `False` like `0` in arithmetic, which is a very useful quirk (for example, `True + True` gives `2`).

4. The **"no value" type:**
    - `None`: a single special value meaning "nothing" or "no value yet" (its type is `NoneType`)

5. **Collection types:**
    - `list`: ordered, mutable sequences, written as `[1, 2, 3]`
    - `tuple`: ordered, immutable sequences like `(1, 2, 3)`
    - `dict` (dictionary): key-value pairs like `{"name": "John", "age": 30}`
    - `set`: unordered collections of unique elements like `{1, "apple", 3.5}`

More information about these types can be found below, and also in the future labs. 

### Numeric types
Numeric data types (`int`, `float`, `complex`) store numbers. They are immutable, meaning that once a number object is created, that specific object cannot be modified. When you perform operations on numbers, Python creates new number objects rather than modifying existing ones.

**Example:**

```python
a = 5               # a is an integer
print(a)

b = 2.5             # b is a float
print(b)

c = 2 + 3j          # c is a complex number
print(c)

earth_mass = 5.722e24   # The earth mass in kg -> a scientific number
print(earth_mass)
```

**Example:**

```python
x = 5       # Creates a number object with value 5
y = x       # y now refers to the same number object
x = x + 1   # This doesn't modify the original 5, it creates a new number 6
# Now x refers to the new 6 object, while y still refers to 5

print(x, y)
```

### Text type

Strings (`str`) store sequences of characters (text). Just like numeric types, they are immutable. Strings can be created using single quotes `'` or double quotes `"`, and joined together with `+`.

```python
text = "Hello"
print(text)
```

Operator `+` concatantes two or more strings:

```python
text2 = text + " World!"   # we have added space to the beginning of World!
print(text2)
```

You can use `*` to repeat a string:

```python
print(2 * text2)
```

### Boolean type

Booleans (`bool`) represent logical values:
- `True`
- `False`

You will most often get booleans as the result of a **comparison**. In addition, any value can be evaluated as a boolean using the `bool()` function: zero and empty values evaluate to `False`, while non-zero and non-empty values evaluate to `True`.

Examples of using `bool()`

```python
print(bool(""))    # empty string
print(bool(0))     # 0 value
print(bool([]))    # an empty list
print(bool(" "))   # One space -> not empty
print(bool(42))    # any value else than 0
```

Examples of comparisons creating Booleans:

```python
print(5 > 3)       # is 5 greater than 3?
print(5 == 3)      # is 5 equal to 3?
```

## Checking types

Python provides two built-in functions to check the type of a value:

- `type()`: returns the type of an object
- `isinstance()`: checks whether an object is of a specific type (or one of several types), returning `True` or `False`

```python
x = 5
print(type(x))                       # <class 'int'>
print(isinstance(x, int))            # True
print(isinstance(x, (int, float)))   # True (checks several types at once)
print(isinstance(x, float))          # False
```

## Converting types

Python can convert data from one type to another. This is called **type conversion**, and it happens in two ways:

* **explicitly**, when you ask for the conversion yourself (also called **casting**), and

* **implicitly**, when Python converts a value automatically (also called **coercion**).


### Explicit conversion (casting)

When we deliberately convert one data type to another using a conversion function, we call it explicit conversion or casting. Common conversion functions include:

|         | string | `int("3.9")`| Error | ValueError: invalid literal for int() with base 10: '3.9'|

|           | string | `float("3.9")` | 3.9 | Converts a string to a floating-point number |

## Precision

Python's handling of numbers, particularly floating-point numbers, can sometimes lead to surprising results due to how numbers are represented in binary. Understanding these limitations is crucial for accurate engineering calculations.

### Floating-point precision

Floating-point numbers in Python (and in most programming languages) use the IEEE-754 double-precision format, which can lead to unexpected results. For instance, `0.1 + 0.2` does not produce exactly `0.3`:

```python
print(0.1 + 0.2)            
print(0.1 + 0.2 == 0.3)     # check if 0.1 + 0.2 is equal to 0.3
```

There are other similar examples:

```python
print(0.3 + 0.3 + 0.3)     
print(0.1 * 3)             
```

This happens because a float is stored in binary using 64 bits (1s and 0s). Just as $1/3$ cannot be written exactly in decimal ($0.3333\ldots$), numbers such as $0.1$ cannot be written exactly in binary, so Python stores the closest value it can. These tiny differences become visible when you add values together or compare them.

A 64-bit float can represent magnitudes from roughly $10^{-308}$ to $10^{308}$, with about 15 to 16 significant decimal digits. The relative rounding error is bounded by the *machine epsilon*: the difference between $1.0$ and the next larger number that can be stored as a float. For Python's `float`, this is $2^{-52} \approx 2.2 \times 10^{-16}$.

```python
import sys

print(format(0.1, ".20f"))      # The value actually stored for 0.1, shown to 20 decimal places
print(sys.float_info.epsilon)   # Machine epsilon, 2**-52
print(sys.float_info.max)       # The largest possible float

print(1.0 + sys.float_info.epsilon == 1.0)       # False: just big enough to change 1.0
print(1.0 + sys.float_info.epsilon / 2 == 1.0)   # True: too small to change 1.0
```

### Integer precision

Integers in Python 3 have unlimited precision (they can be as large as your computer's memory allows), unlike floats:

```python
# Very large integers work fine
print(2 ** 100)             # 1267650600228229401496703205376

# But mixing them with floats converts the result to a float, losing digits
print(2 ** 100 + 0.1)       # Only about 16 significant digits are kept

# An extremely large integer
big_number = 2 ** 10000
print(len(str(big_number))) # It has 3011 digits, and Python handles it without problem
```

Mixing floats with extremely large integers can even cause an `OverflowError`:

```python
print(big_number + 0.1)     # OverflowError: int too large to convert to float
```

### Dealing with precision issues

Several approaches exist to handle precision issues. <br> 
The most relevant one for us is using `math.isclose()` to compare floats instead of `==`. With `math.isclose()` we check whether two variables are **close enough** within a tolerance. The tolerance can be set through an optional argument `abs_tol` with the default value of $10^{-9}$.

```python
from math import isclose

print(0.1 + 0.2 == 0.3)                      # False
print(isclose(0.1 + 0.2, 0.3))               # True (with the default tolerance of 1e-9)
```

A relative tolerance does not work when comparing with zero, so give a smaller absolute tolerance:

```python
print(isclose(1e-20, 0.0))                   # False!
print(isclose(1e-20, 0.0, abs_tol=1e-12))    # True
```
