## Arithmetic Operations

The basic arithmetic operations in Python are listed below.

| **Operation** | **Operator** |     
| :- | :- | 
| Addition | $+$ |
| Subtraction | $-$ | 
| Multiplication | $*$ |
| Float Division | $/$ |
| Floor Division | $//$ |
| Exponentiation | $**$ |
| Modulo (remainder) | \% |

### Output type

In general, 
* The output type depends on the type of the operands and the type of operation. 
* But, the outcome of float division is always a float.

Through the following experiments, you can verify these points. 

* Assume variables $x = 5$ and $y = 8$ to be integers. Then, investigate the outcome of the above operations and their type.
* Repeat the above, assuming $x = 5.1$ and $y=8$, and then $x = 5.1$ and $y=8.0$.
* Repeat the above, assuming $x = 5.2$ and $y=8$, and then $x = 5.2$ and $y=8.0$. 

**Task:** What are your conclusions? Do you observe any unexpected result? Motivate your observations. 

```python
# Assign variables of different type
x = 5
y = 8

# arithmetic operations
s = x + y    # addition
b = x - y    # subtraction
m = x * y    # multiplication
d = x / y    # float division
f = x // y   # floor division
e = x ** y   # exponentiation
o = x % y    # remainder 

#print the outcome
print(f'{x} + {y} = {s}; output type = {type(s)}')
print(f'{x} - {y} = {b}; output type = {type(b)}')
print(f'{x} * {y} = {m}; output type = {type(m)}')
print(f'{x} / {y} = {d}; output type = {type(d)}')
print(f'{x} // {y} = {f}; output type = {type(f)}')
print(f'{x} ** {y} = {e}; output type = {type(e)}')
print(f'{x} % {y} = {o}; output type = {type(o)}')
```

## Absolute value 

For absolute value $|x|$, there is a native Python function: `abs(x)`. 

* For a integer/float number `x`, `abs(x)` returns the absolute value of `x`.
* For complex number `x`, `abs(x)` returns the magnitude of `x`.

See the following example:

```python
a1 = -3.65         # float
a2 = -3.65 + 2.0j  # complex

print(f"|a1| = {a1}")
print(f"|a2| = {a2}")
print(f"|a1 + a2| = {a1 + a2}")
```

## Mathematics Libraries

The arithmetic operations and `abs(x)` discussed above, are built into Python. However, we often encounter other mathematical functions, such as $\sin(x)$, $\sin^{-1}(x)$, $\log(x)$, $\exp(x)$, etc., in mathematical expressions. These functions can be **imported from Python libraries**.

- For mathematical functions with **scalar inputs and outputs**, we can use the `math` and `cmath` libraries, as discussed below.
- For mathematical functions that operate on **arrays**, we will use the `numpy` library, which will be introduced in future lectures and computer labs.

Focusing on scalar inputs and outputs, the choice between `math` and `cmath` mainly depends on whether we are working with real or complex numbers:

* `math`: Provides mathematical functions for real numbers, including integers (int) and floating-point numbers (float).
* `cmath`: Provides mathematical functions for complex numbers (complex).

| Input type | Library | List of functions |
| :- | :- | :- |
| Real number | `math` | [math functions](https://docs.python.org/3/library/math.html)
| Complex number | `cmath` | [cmath functions](https://docs.python.org/3/library/cmath.html#module-cmath)

In the links provided, you can see different functions including trigonometric, logarithmic, hyperbolic, etc. 

To use a function, we need to import the corresponding library at the beginning of a Python script. The imported libraries can also be given shorter names for convenience.

```python
import math as mt
import cmath as cmt
```

**Example:** For a real (float or integer) input, we can call functions from either of the these two libraries. Note that the format of the output can be different depending on the input type and operations.  

```python
a = 3.6   # float 

print(mt.sin(a))   # math function  -> real output
print(cmt.sin(a))  # cmath function -> complex output
```

**Example:** For a complex number input, we can only use functions from `cmath`:

```python
z = 3.6 + 2.5j   #complex number input

print(cmt.sin(z))    #cmath function
```

If you try `print(mt.sin(z))`, you will get a `TypeError`.

### Some notes on the mathematical functions

* For the trigonometric functions such as $\sin(x)$, $\cos(x)$, etc imported from the above libraries, $x$ is in radians. You can convert degrees and radians to each other using `math.radians(x)` and `math.degrees(x)`, respectively.

```python
print(f"sin(30 deg) = {mt.sin(mt.radians(30))}")
print(f"tan(45 deg) = {mt.tan(mt.radians(45))}")
```

* Be careful about $\ln(x)$ that is `math.log(x)`, and base-10 $log(x)$ that is `math.log10(x)`.

```python
print(f" ln(100) = {mt.log(100)}")
print(f" log(100) = {mt.log10(100)}")
```

* Providing inadmissible values to a particular mathematical function will generate an error. Therefore, a prior knowledge of the admissible domain a mathematical function is important. 

For instance, we know for $\sin^{-1}(x) = \arcsin(x)$, we should have $-1 \leq x\leq 1$. Trying an $x$ outside of this range will generate `ValueError: math domain error`. You can try it below. 

```python
x = 0.5  #try a value greater than 1 or smaller than -1
print(f"arcsin({x}) = {mt.asin(x)} radians.")
```

In particular, pay attention to the square root of a negative value. There is no issue if the `sqrt()` function is imported from `cmath`, but there will be a domain error if the same function is imported from `math`. 

```python
b = -1.5 

print(cmt.sqrt(b))  # using cmath.sqrt -> fine
#print(mt.sqrt(b))  # using math.sqrt -> math domain error
```

## Constants
The above libraries have values for some of the common consttants, see the following table (you will learn about `numpy` in next notebooks):

| Constant | Python value | Description |
| :-- | :-- | :-- |
| $\pi$ | `math.pi`, `numpy.pi` | Pi, approximately 3.141592653589793 |
| $e$ | `math.e`, `numpy.e` | Euler's number (Napier's constant), approximately 2.718281828459045; base of the natural logarithm |
| $\infty$ | `math.inf`, `numpy.inf` | Floating-point representation of positive infinity |
| `NaN` (Not a Number) | `math.nan`, `numpy.nan` | Special floating-point value representing an undefined or unrepresentable numerical result |

**Note:** `inf` and `nan` are special floating-point values.
For example, `numpy.sqrt(-1.0)` returns `nan` (with a warning), while
`numpy.log(0.0)` returns `-inf` (also with a warning).



Similar constants from `math` and `numpy` have the same values. 

```python
print(f'math.pi = {mt.pi}')
print(f'math.e = {mt.e}')
print(f'math.inf = {mt.inf}')
print(f'math.nan = {mt.nan}')
```

To check whether a value is `nan` or `inf`, use the following functions:

* `math.isnan()` or `numpy.isnan()`: Check whether a value is `nan`.
* `math.isinf()` or `numpy.isinf()`: Check whether a value is positive or negative infinity.
* `math.isfinite()` or `numpy.isfinite()`: Check whether a value is finite (neither `nan` nor `inf`).

```python
print(mt.isinf(mt.inf))   
print(mt.isinf(-mt.inf)) 
print(mt.isnan(mt.inf))  
print(mt.isinf(10.0))      
print(mt.isfinite(10.0))    
print(mt.isfinite(mt.inf)) 
```

## Mathematical Expressions

You can use Python built-in operations and functions, as well as functions imported from libraries like `math`, `cmath`, and later `numpy` to translate mathematical expressions to Python language. <br>

When doing so, the most important thing is to be aware of the **precedence order of operations**, that is: <br> 
    ( ) $\quad \to \quad$ ** $\quad \to \quad$ unary - $\quad \to \quad$ * / // \% $\quad \to \quad$ + - <br>
    Parentheses $\to$ power $\to$ negative sign $\to$ multiplication, float division, floor division, remainder $\to$ addition, subtraction


**Example:** See the difference between $-2^4$, $(-2)^4$, and $-2^{-4}$:

```python
print(-2**4)
print((-2)**4)
print(-2**(-4))
```

**Example:** See the difference between $2^{3^2}$, $(2^3)^2$:

```python
print(2**3**2)
print((2**3)**2)
```

Now, let's evaluate a few mathematical expressions in Python. 

Evaluate $y = \sqrt{x^2 + \exp(1-x)\sin(x)}$ at a given $x$:

```python
x = 2.2

y = mt.sqrt(x**2 + mt.exp(1-x) * mt.sin(x))
print(y)
```

Evaluate $y$ defined in the following expression for given values of $x$ (in degrees) and $t$. Print the output $y$ with 4 decimal places in standard and scientific formats. 

$$
y = t\sin(2x)+\cos^{-1}(\frac{2}{t+1}) + \ln(t-2) + e^{1-t^2} + \sqrt{t}
$$

Note: The value of $t$ should be greater than 2 (why?). 

```python
x = 22.3
t = 5.0 

y = (t * mt.sin(2*x) + mt.acos(2/(t+1)) 
    + mt.log(t-2) + mt.exp(1-t**2) + mt.sqrt(t))
    
print(f"y = {y:.4f}")  # standard float, 4 decimal places
print(f"y = {y:.4e}")  # scientific float, 4 decimal places
```
