## Printing to screen

As you have already seen, the most basic way of displaying an output is to use the `print` function:

```python
print("This displays in your terminal")
```

`print` can display several items at once, including numbers. If you separate the items with commas, they are printed with a space between them:

```python
print("This displays in your", "terminal")
print("The answer is", 42, "not", 52)
```

There are two ways for formatting the outputs: 
* using place holders `%d`, `%g`, `%f`, etc. 
* using f-strings

In this lab, we only cover the latter which is the most recent and standard way of formatting in Python.                   

### f-strings
Often, we want to include the value of a variable within some text, and control how that value is displayed. The easiest way to do this is to use an **f-string**, which is a string with the letter `f` placed before the opening quote. Anything inside curly braces `{}` is evaluated as standard code, and the result is placed in the string. What is provided in `{}` can be a single variable or a calculation/operation applied to variable(s). 

Consider the following simple example:

```python
temperature = 15.0

print(f"The temperature is {temperature} degrees Celsius")
print(f"In Kelvin, this is {temperature + 273.15}")
```

Be careful not to forget the `f`. Without it, no error is raised, but the curly braces are simply printed as ordinary text:

```python
print("The temperature is {temperature} degrees Celsius")
```

### Format codes
We can control how a value is displayed by adding a colon and a **format code** inside the curly braces, for example `{temperature:d}`, or in this example `{a:d}`. The most common format codes are shown in the following table.

| Data type | Formatting type | Format code |
| :- | :- | :- |
| integer | Standard    | **:d** |
| integer | Hexadecimal | **:x** |
| float   | Standard  | **:f** |
| float   | General-format    | **:g** |
| float   | Scientific   | **:e** |
| string  | Standard | **:s** |

For a full ist of all possible formatting codes which are a part of this f-string mini-language can be found in the Python documentation:

https://docs.python.org/3/library/string.html#formatspec

**Example**: Print an integer number using various formatting:

```python
a = -159   # an integer

print(f"a with no format code = {a}")
print(f"a as integer = {a:d}")
print(f"a as float = {a:f}")
print(f"a as general float = {a:g}")
print(f"a as scientific float = {a:e}")
print(f"a as hexadecimal = {a:x}")
```

**Example**: Print a float number using various formatting:

```python
b = -159.843962  # a float number

print(f"b with no format code = {b}")
print(f"b as float = {b:f}")
print(f"b as general float = {b:g}")
print(f"b as scientific float = {b:e}")
print(f"b cast to an integer = {int(b):d}")
```

**Note**: The integer format codes only work on integers, so `{b:d}` raises a `ValueError` when `b` is a float. This is why `b` is cast using `int()` above. Remember that casting truncates the value rather than rounding it.

We can also specify the **width** and the **precision**, i.e. the number of digits after the decimal point for a `float`.

For instance, for `{b:12.2f}` or `{b:12.2e}`:
* `12`: is the minimum width of the entire formatted value. If the value takes fewer characters, spaces will be added on the left for padding. The width is optional, and can be left out.
* `2`: sets the precision to 2, meaning that 2 digits will be displayed after the decimal point. The value is automatically rounded.

```python
b = -159.843962  # a float number

print(f"b to 2 decimal places = {b:.2f}")
print(f"b to 3 decimal places = {b:.3f}")
print(f"b in scientific notation = {b:.2e}")
print(f"b in scientific notation = {b:.3e}")
print(f"b with a width of 12 = {b:12.2f}")
```

**Note**: You may come across older code that formats strings using the `%` operator, such as `print("b = %.2f" % b)`. This does the same job, but we will use f-strings throughout this course.

## Runtime Input

To receive keyboard inputs from your terminal after running a script, we use Python's built-in `input()` function. This function reads the user's input and returns it as a **string**. <br> 

Run the following exmaple, and try different types of inputs: `int`, `float`, and `str`.

```python
a = input("Enter your input: ")

print(f"Received input is: {a} which is of type {type(a)}")
```

If we want text, this will work perfectly. However, the result is still a string even if the user types a number, so we need to **convert** it to the right type before doing any maths.
* To convert the input to an integer, use `int()`. But, as explained in [the variables notebook (explicit conversion)](../variables.ipynb), something like `int("3.2")` leads to an error, while `int("3")` runs perfectly fine. 
* To convert the input to a float, use `float()`.

```python
a = input("Enter a number: ")

print(f"Original type: {type(a)}")

a1 = float(a)   # conversion from string to float
print(f"Converted value: {a1}")
print(f"Converted type: {type(a1)}")
```

Now that the value has been converted to a float, we can use it in mathematical operations. For instance:

```python
print(f"a1 squared = {a1**2:.2f}")
print(f"1/a1 = {1/a1:.4f}")
```

### Multiple inputs

If we have multiple inputs, there are two options:
* **Option 1**: Repeat the `input` command multiple times, each time reading the value of one of the inputs.
* **Option 2**: Use the `input` command only once, but use `split()`. 

```python
# Option 1
a = input("Enter value of a: ")
b = input("Enter value of b: ")
print(f"a = {a}, b = {b}")
```

For Option 2, use *space* to separate the input values. Both values are still strings, so they must be converted before doing any maths:

```python
# Option 2
a, b = input("Enter values of a and b: ").split()
print(f"a = {a}, b = {b}")

a = float(a)
b = float(b)
print(f"a + b = {a + b}")
```

## Summary

* The `print()` function displays output on the screen.
* f-strings place values directly in a string using curly braces, as in `f"x = {x}"`.
* Format codes control how values are displayed. For example, `{x:.2f}` displays a float to 2 decimal places, and `{x:.3e}` uses exponential notation with 3 decimal places.
* The `input()` function always returns a string, so numbers must be converted using `int()` or `float()` before they can be used in calculations.
* Several values can be read in one line by using `input().split()`.
